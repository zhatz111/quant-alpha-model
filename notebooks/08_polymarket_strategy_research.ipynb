### Base strategy to build on

**Phase 1: Initial Market Entry**
- If either position is 0, build an initial balanced position
- Check to see if we own any UP, then check to see if we own any DOWN
    - buy UP first then DOWN if neither is owned

**Phase 2: Hedging**
- Check the imbalance in positions, if more UP then DOWN, buy DOWN and vice versa
- Before buying either side check to see if potential pair cost calculated as the `as_up_price` plus the `avg_down_price` and vice versa are greater than the `target_pair_cost`.
- Calculate the size to buy that is needed to neutralize the position

**Phase 3: Pure Arbitrage**
- If the `ask_up_price` plus `ask_down_price` are less than 1 then buy both legs at same time (risky due to both sides not filling).

**Phase 4: Imbalance Targetting**
- If we are approaching the imabalance limit between our positions determine which side is most imbalanced and try to buy that side (risky cause it might trade at bad conditions in market when we should not actually trade).

**Phase 5: Standard Arbitrage**
- Check if `ask_up_price` + `avg_down_price` or `ask_down_price` + `avg_up_price` are below the `target_pair_cost`, if they are then trade that side.
- If that criteria is not met then use an imbalance signal like the relative imbalance. If it is beyond a certain threshold for one side, then trade that side.

**Other Factors**
- Position delta constraint: if the position imbalance is greater than the allowed max imabalance then stop trading (risky because this might stop all trading for the rest of the market).
